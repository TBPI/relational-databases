## Running SQL (PostegreSQL) in Jupyter Notebook

We want to do the exercises in the notebook.
For this, we need to connect to our PostgreSQL account without presenting the password to GitHub.
This can be done via the `SQL Magic`
```
postgresql+psycopg://postgres@localhost:5432/postgres
```
where
- `postgresql+psycopg://` is the driver,
- `postgres` is the user,
- `localhost` is the host,
- `5432` is the port,
- `postgres` is the database (by default there is a postgres database).

````python
%load_ext sql
%sql postgresql+psycopg://postgres@localhost:5432/postgres
````

````python
%sql SELECT version();
````

We can check the current database and the current user

````python
%%sql
SELECT current_database(), current_user;
````

### Shortcuts for databases

Using the package `pgspecial`, we can run
- `\dt` to list all tables in a given database,
- `\l` to list all databases,
- `\d table_name` to show all columns and types of a table.

We can connect to another database (e.g. test1) as follows.

````python
%sql postgresql+psycopg://postgres@localhost:5432/test1 --alias test1
%sql postgresql+psycopg://postgres@localhost:5432/bikes --alias bikes
````

Check the open connections.

````python
%sql --connections
````

We can use the alias to switch between open connections

````python
%sql test1
````

In the end, we want to close all the open connections.

````python
%sql --close bikes
%sql --close test1
````
