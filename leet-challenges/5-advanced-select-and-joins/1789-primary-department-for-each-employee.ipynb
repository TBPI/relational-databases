# Exercise 31: Primary Department for Each Employee

**Source:** [LeetCode 1789](https://leetcode.com/problems/primary-department-for-each-employee/description/?envType=study-plan-v2&envId=top-sql-50) (Easy).

Table: `Employee`
```
+---------------+---------+
| Column Name   |  Type   |
+---------------+---------+
| employee_id   | int     |
| department_id | int     |
| primary_flag  | enum    |
+---------------+---------+
```
- `(employee_id, department_id)` is the primary key.
- `department_id` is the department where the employee `employee_id` belongs to. An employee can belong to several departments.
- `primary_flag` is an `ENUM` with values `Y` and `N`. If an employee belongs only to one department the flag is on `N`. Otherwise, the flag is put on `Y` for the primary department of the employee and `N` for all others.

**Task.** Return **in any order** the `employee_id` and the `department_id` of the employee's primary department. (It is the employee's unique department if the employee only belongs to one).

#### Fill the database with content

The `ENUM` `prim` needs to be created

````python
%sql CREATE TYPE prim AS ENUM ('Y', 'N');
````

Now, create the database.

````python
%%sql
DROP TABLE IF EXISTS Employee;

CREATE TABLE Employee (
    employee_id INTEGER NOT NULL,
    department_id INTEGER NOT NULL,
    primary_flag prim NOT NULL,
    PRIMARY KEY (employee_id, department_id)
);

INSERT INTO Employee (employee_id, department_id, primary_flag)
VALUES
(1, 1, 'N'),
(2, 1, 'Y'),
(2, 2, 'N'),
(3, 3, 'N'),
(4, 2, 'N'),
(4, 3, 'Y'),
(4, 4, 'N');

SELECT * FROM Employee;
````

#### Solve the exercise

**Task.** Return **in any order** the `employee_id` and the `department_id` of the employee's primary department. (It is the employee's unique department if the employee only belongs to one).

````python
%%sql
SELECT E.employee_id, E.department_id
FROM Employee E INNER JOIN (
    SELECT
        employee_id, COUNT(department_id) AS dep_numbers
    FROM Employee
    GROUP BY employee_id) D
ON E.employee_id = D.employee_id
WHERE primary_flag = 'Y' OR D.dep_numbers = 1
````

#### Explanation
- With a subquery create a table that counts for every `employee_id` the number of departments.
- Join this table with the table `Employee` on `employee_id`.
- Return the columns with count of departments equal to `1` or `primary_flag` toggled to `Y`.
