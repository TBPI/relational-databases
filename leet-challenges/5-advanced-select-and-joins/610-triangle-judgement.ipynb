# Exercise 32: Triangle judgement

**Source:** [LeetCode 610](https://leetcode.com/problems/triangle-judgement/description/?envType=study-plan-v2&envId=top-sql-50) (Easy).

Table: `Triangle`
```
+-------------+------+
| Column Name | Type |
+-------------+------+
| x           | int  |
| y           | int  |
| z           | int  |
+-------------+------+
```
 - `(x, y, z)` is the primary key describing the length of three line segments.

 **Task.**
 Determine whether the three line segments can be arranged into a triangle by adding a column `triangle` with `Yes` or `No`.

Connect to the database `postgres` using the username `postgres`.

````python
%load_ext sql
%sql postgresql+psycopg://postgres@localhost:5432/postgres
````

### Create new database for the exercise

Make sure first that a database with the same name is removed in case it already exists.

````python
%sql DROP DATABASE IF EXISTS exercise_32 WITH (FORCE);
%sql CREATE DATABASE exercise_32;
````

Connect to the database.

*Caution.* No semicolon at the end otherwise it belongs to the name of the database.

````python
%sql postgresql+psycopg://postgres@localhost:5432/exercise_32 --alias exercise_32
````

#### Fill the database with content

````python
%%sql
DROP TABLE IF EXISTS Triangle;

CREATE TABLE Triangle (
    x INTEGER NOT NULL,
    y INTEGER NOT NULL,
    z INTEGER NOT NULL,
    PRIMARY KEY (x,y,z)
);

INSERT INTO Triangle (x,y,z)
VALUES
(13, 15, 30),
(10, 20, 15);
````

#### Solve the exercise

**[Exercise task]**

````python
%%sql
SELECT x, y, z, CASE
    WHEN x + y > z AND x + z > y AND z + y > x AND x > 0 AND y > 0 AND z > 0 THEN 'Yes'
    ELSE 'No'
    END
    AS triangle
FROM Triangle
````

#### Explanation
- Make a case distinction for the additional column `Triangle`. We need the triangle inequality to be strict for all three possibilites to have `YES` as outcome.
