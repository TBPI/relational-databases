# Exercise 33: Consecutive Numbers

**Source:** [LeetCode 180](https://leetcode.com/problems/consecutive-numbers/description/?envType=study-plan-v2&envId=top-sql-50) (Medium).

Table: `Logs`
```
+-------------+---------+
| Column Name | Type    |
+-------------+---------+
| id          | int     |
| num         | varchar |
+-------------+---------+
```
- `id` is the autoincrementing primary key (`SERIAL`).
- `num` is an number in varchar form.

**Task.**
Return all numbers that appear **three times** (or more) **consecutively** in **any order**.

Connect to the database `postgres` using the username `postgres`.

#### Fill the database with content

````python
%%sql

DROP TABLE IF EXISTS Logs;

CREATE TABLE Logs (
    id SERIAL PRIMARY KEY,
    num VARCHAR
);

INSERT INTO Logs (num) VALUES
('1'),
('1'),
('1'),
('2'),
('1'),
('2'),
('2');

SELECT * FROM Logs;
````

#### Solve the exercise

**Task.**
Return all numbers that appear **three times** (or more) **consecutively** in **any order**.

````python
%%sql
SELECT
    DISTINCT num1 AS ConsecutiveNums
FROM
    (SELECT
        P.num AS num1, S.num AS num2, T.num AS num3
    FROM
        Logs P
    LEFT JOIN Logs S
    ON
        P.id = S.id - 1
    LEFT JOIN Logs T
    ON
        P.id = T.id - 2
    )
WHERE
    num1 = num2 AND num2 = num3
````

#### Explanation

- First create a table that contains three consecutive numbers from the table using two joins.
- Then take only the numbers where the columns `num1`, `num2` and `num3` coincide.
