+---------------+---------+

- `id` is the primary key.
- The table records date and success/failure of an incoming transaction.
- The `state` is an `ENUM` with categories (`approved`, `declined`).

**Task.**

Return the result in any order.

#### Fill the database with content

````python
%%sql
DROP TABLE IF EXISTS Transactions;

-- Create the ENUM type
DROP TYPE IF EXISTS STATUS;
CREATE TYPE STATUS AS ENUM ('approved', 'declined');

-- Create table Transactions
CREATE TABLE Transactions (
    id INTEGER PRIMARY KEY,
    country VARCHAR NOT NULL,
    state STATUS,
    amount INTEGER,
    trans_date DATE
);

INSERT INTO Transactions (id, country, state, amount, trans_date)
VALUES
(121, 'US', 'approved', 1000, '2018-12-18'),
(122, 'US', 'declined', 2000, '2018-12-19'),
(123, 'US', 'approved', 2000, '2019-01-01'),
(124, 'DE', 'approved', 2000, '2019-01-07');
````

#### Solve the exercise

**Task.**
Find for each month and country the number of transactions `trans_count`, the total amount of transactions `trans_total_amount`, the number of approved transactions `approved_count`, and the amount of approved transactions `approved_total_amount`.
Return the result in any order.

````python
%%sql
SELECT
    TO_CHAR(DATE_TRUNC('month', trans_date), 'YYYY-MM') AS month,
    country,
    COUNT(*) AS trans_count,
    COUNT(CASE WHEN state = 'approved' THEN 1 END) AS approved_count,
    SUM(amount) AS trans_total_amount,
    COALESCE(SUM(CASE WHEN state = 'approved' THEN amount END),0) AS approved_total_amount
FROM
    Transactions
GROUP BY
    month, country;
````
