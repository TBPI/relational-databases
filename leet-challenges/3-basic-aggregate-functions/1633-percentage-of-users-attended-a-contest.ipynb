# Exercise 18: Percentage of Users Attended a Contest

**Source** [LeetCode 1633](https://leetcode.com/problems/percentage-of-users-attended-a-contest/description/?envType=study-plan-v2&envId=top-sql-50) (Easy).

Table: `Users`
```
+-------------+---------+
| Column Name | Type    |
+-------------+---------+
| user_id     | int     |
| user_name   | varchar |
+-------------+---------+
```
- `user_id` is the primary key.

Table: `Register`
```
+-------------+---------+
| Column Name | Type    |
+-------------+---------+
| contest_id  | int     |
| user_id     | int     |
+-------------+---------+
```
- `(contest_id, user_id)` is the primary key.
- A row describes the registration of a user `user_id` to the contest `contest_id`.

**Task**.
Find the percentage `percentage` of users (among the `Users` table) registered to a context, **rounded to 2 decimals**.
Return first in **descending order** of the `percentage` and second **ascending order** by `contest_id`.

#### Fill the database with content

````python
%%sql
DROP TABLE IF EXISTS Users;
DROP TABLE IF EXISTS Register;

CREATE TABLE Users (
    user_id INTEGER PRIMARY KEY,
    user_name VARCHAR NOT NULL
);

INSERT INTO Users (user_id, user_name)
VALUES
(6, 'Alice'),
(2, 'Bob'),
(7, 'Alex');

CREATE TABLE Register (
    contest_id INTEGER NOT NULL,
    user_id INTEGER NOT NULL,
    PRIMARY KEY (contest_id, user_id),
    FOREIGN KEY (user_id) REFERENCES Users(user_id)
);

INSERT INTO Register (contest_id, user_id)
VALUES
(215, 6),
(209, 2),
(208, 2),
(210, 6),
(208, 6),
(209, 7),
(209, 6),
(215, 7),
(208, 7),
(210, 2),
(207, 2),
(210, 7);

SELECT * FROM Register;
````

#### Solve the exercise

**Task**.
Find the percentage `percentage` of users (among the `Users` table) registered to a context, **rounded to 2 decimals**.
Return first in **descending order** of the `percentage` and second **ascending order** by `contest_id`.

````python
%%sql
SELECT
    contest_id, ROUND(100 * COUNT(user_id)::NUMERIC/(SELECT COUNT(*) FROM Users), 2) AS percentage
FROM
    Register
GROUP BY 
    contest_id
ORDER BY
    percentage DESC,
    contest_id ASC;
````

#### Explanation

- The `Users` table is only needed to have the total number of users. One can use it as a subquery `SELECT COUNT (*) FROM Users` inside the calculation of the `percentage`.
- Group the `Register` table by `contest_id` calculate the percentage using `COUNT(user_id)`, rounded to two decimals with `ROUND(_, 2)`.
- Do `ORDER BY` with inputs: first descending `percentage`, then ascending `contest_id`.
