#### Fill the database with content

````python
%%sql
DROP TABLE IF EXISTS Project;
DROP TABLE IF EXISTS Employee;

CREATE TABLE Employee (
    employee_id INTEGER PRIMARY KEY,
    name VARCHAR NOT NULL,
    experience_years INTEGER
);

CREATE TABLE Project (
    project_id INTEGER NOT NULL,
    employee_id INTEGER NOT NULL,
    PRIMARY KEY (project_id, employee_id),
    FOREIGN KEY (employee_id) REFERENCES Employee(employee_id)
);

INSERT INTO Employee (employee_id, name, experience_years)
VALUES
(1, 'Khaled', 3),
(2, 'Ali', 2),
(3, 'John', 1),
(4, 'Doe', 2);

INSERT INTO Project (project_id, employee_id)
VALUES
(1, 1),
(1, 2),
(1, 3),
(2, 1),
(2, 4);

SELECt * FROM Project
````

#### Solve the exercise

**Task.**
For each project find the **mean experience** among all users working on the project **rounded to 2 decimals**. Return the result in **any order**.

````python
%%sql
SELECT P.project_id, ROUND(AVG(E.experience_years),2) AS average_years
FROM
    Project P LEFT JOIN Employee E
ON
    P.employee_id = E.employee_id
GROUP BY project_id;
````

#### Explanation

- First `LEFT JOIN` the tables on `employee_id`.
- Then group by the `project_id` in taking the `AVG()`. Don't forget to `ROUND(_, 2)`.
