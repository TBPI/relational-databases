Table: `Tweets`
```
+----------------+---------+
| Column Name    | Type    |
```

- `tweet_id` is the primary key (column with unique values),

Write a solution to find the IDs of the *invalid tweets*. The tweet is **invalid** if the number of characters used in the content of the tweet is strictly greater than 15.

#### Fill the database with content

````python
%%sql
DROP TABLE IF EXISTS Tweets;

CREATE TABLE Tweets (
    tweet_id INTEGER PRIMARY KEY,
    content VARCHAR
);
INSERT INTO Tweets (tweet_id, content)
VALUES
(1, 'Let us Code'),
(2, 'More than fifteen chars are here!');
SELECT * FROM Tweets
````

#### Solve the exercise

**Exercise.**
Write a solution to find the IDs of the *invalid tweets*. The tweet is **invalid** if the number of characters used in the content of the tweet is strictly greater than 15.

Return the result table in **any order**.

````python
%sql SELECT tweet_id FROM Tweets WHERE LENGTH(content) > 15;
````

#### Explanation

- `LENGTH(Content)` measures the length of the string in the given row. This should be strictly greater than 15 for an *invalid tweet*.
