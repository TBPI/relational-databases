# Exercise 2: Find Customer Referee

**Source:** [LeetCode 584](https://leetcode.com/problems/find-customer-referee/) (Easy).

Table: `Customer`

```
+-------------+---------+
| Column Name | Type    |
+-------------+---------+
| id          | int     |
| name        | varchar |
| referee_id  | int     |
+-------------+---------+
```

- `id` is the primary key.
- `referee_id` is the id of the customer who referred them (`NULL` if nobody did).

**Task.** List the names of all customers who were not referred by the customer with id 2, including those not referred at all, in any order.

Connect to the database `postgres` using the username `postgres`.

````python
%load_ext sql
%sql postgresql+psycopg://postgres@localhost:5432/postgres
````

### Create new database for the exercise

Make sure first that a database with the same name is removed in case it already exists.

````python
%sql DROP DATABASE IF EXISTS exercise_2 WITH (FORCE);
%sql CREATE DATABASE exercise_2;
````

Connect to the database.

*Caution.* No semicolon at the end otherwise it belongs to the name of the database.

````python
%sql postgresql+psycopg://postgres@localhost:5432/exercise_2 --alias exercise_2
````

#### Fill the database with content

````python
%%sql
DROP TABLE IF EXISTS Customer;

CREATE TABLE Customer (
    id INTEGER PRIMARY KEY,
    name VARCHAR NOT NULL,
    referee_id INTEGER
);
INSERT INTO Customer (id, name, referee_id)
VALUES (1, 'Will', NULL),
(2, 'Jane', NULL),
(3, 'Alex', 2),
(4, 'Bill', NULL),
(5, 'Zack', 1),
(6, 'Mark', 2);
SELECT * FROM Customer;
````

#### Solve the exercise

**Task.**
List the names of all customers who were not referred by the customer with id 2, including those not referred at all, in any order.

````python
%sql SELECT name FROM Customer WHERE referee_id != 2 OR referee_id IS NULL;
````

#### Explanation
One needs to select the column `name` only.
The `WHERE`-clause ensures that
- that the `referee_id` is `NULL` (the customer is not referred at all)
- or, in the case it is not `NULL`, the `referee_id` is not 2.

#### Close connection

````python
%sql --close exercise_2
%sql postgresql+psycopg://postgres@localhost:5432/postgres
%sql DROP DATABASE IF EXISTS exercise_2;
````
