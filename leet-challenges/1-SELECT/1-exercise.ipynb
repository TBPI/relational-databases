Table: `Products`
```
+-------------+---------+
| Column Name | Type    |
+-------------+---------+
| product_id  | int     |

+-------------+---------+
```
- `product_id` is the primary key.
- `low_fats` and `recyclable` are `'Y'`/`'N'` flags.

**Task.** List the ids of all products that are both low fat and recyclable, in any order.

### Create the database for the exercise

````python
%sql DROP DATABASE IF EXISTS exercise_1 WITH (FORCE);
%sql CREATE DATABASE exercise_1;
````

Connect to the database.

````python
%sql postgresql+psycopg://postgres@localhost:5432/exercise_1 --alias exercise_1
````

We fill the database with content.

We need to first create a new type for categorical variables.

````python
%%sql
CREATE TYPE yesno AS ENUM ('Y', 'N');
````

Now we can fill the database.

````python
%%sql
DROP TABLE IF EXISTS Products;

CREATE TABLE Products (
    product_id INTEGER PRIMARY KEY,
    low_fats yesno NOT NULL,
    recyclable yesno NOT NULL
);
INSERT INTO Products (product_id, low_fats, recyclable)
VALUES (0, 'Y', 'N'),
(1, 'Y', 'Y'),
(2, 'N', 'Y'),
(3, 'Y', 'Y'),
(4, 'N', 'N');
SELECT * FROM Products;
````

#### Solve the exercise.

**Task.**
List the ids of all products that are both low fat and recyclable, in any order.

````python
%%sql
SELECT product_id FROM Products
WHERE low_fats = 'Y' AND recyclable = 'Y';
````

We want to column `product_id` from the table Products.

The `WHERE`-clause ensures that we can impose conditions on which columns to use.

We want the two conditions `low_fats = 'Y'` and `recyclable = 'Y'` to hold which is ensured by `AND` connector.

````python
%sql --close exercise_1
%sql postgresql+psycopg://postgres@localhost:5432/postgres
%sql DROP DATABASE IF EXISTS exercise_1;
````
