# Exercise 3: Big Countries

**Source:** [LeetCode 595](https://leetcode.com/problems/big-countries/) (Easy).

Table: `World`
```
+-------------+---------+
| Column Name | Type    |
+-------------+---------+
| name        | varchar |
| continent   | varchar |
| area        | int     |
| population  | int     |
| gdp         | bigint  |
+-------------+---------+
```
- `name` (of the country) is the primary key.
- `area` is in km², `gdp` in dollars.

**Task.** A country counts as *big* if its area is at least 3,000,000 km² or its population is at least 25,000,000. List name, population and area of all big countries, in any order.

Connect to the database `postgres` using the username `postgres`.

````python
%load_ext sql
%sql postgresql+psycopg://postgres@localhost:5432/postgres
````

### Create new database for the exercise

Make sure first that a database with the same name is removed in case it already exists.

````python
%sql DROP DATABASE IF EXISTS exercise_3 WITH (FORCE);
%sql CREATE DATABASE exercise_3;
````

Connect to the database.

*Caution.* No semicolon at the end otherwise it belongs to the name of the database.

````python
%sql postgresql+psycopg://postgres@localhost:5432/exercise_3 --alias exercise_3
````

#### Fill the database with content

````python
%%sql
DROP TABLE IF EXISTS World;

CREATE TABLE World (
    name VARCHAR PRIMARY KEY,
    continent VARCHAR NOT NULL,
    area INTEGER,
    population INTEGER,
    gdp BIGINT
);
INSERT INTO World (name, continent, area, population, gdp)
VALUES 
('Afghanistan', 'Asia', 652230, 25500100, 20343000000),
('Albania', 'Europe', 28748, 2831741, 12960000000),
('Algeria', 'Africa', 2381741, 37100000, 188681000000),
('Andorra', 'Europe', 468, 78115, 3712000000),
('Angola', 'Africa', 1246700, 20609294, 100990000000);
SELECT * FROM World;
````

#### Solve the exercise

**Task.**
A country counts as *big* if its area is at least 3,000,000 km² or its population is at least 25,000,000. List name, population and area of all big countries, in any order.

````python
%sql SELECT name, population, area FROM World WHERE area >= 3000000 OR population >= 25000000;
````

#### Explanation

- The three columns to select are `name`, `population`, `area`.
- At least one of the two conditions must hold `area >= 3000000` or `population >= 25000000` using the `OR` connector in the `WHERE` clause.

#### Close connection

````python
%sql --close exercise_3
%sql postgresql+psycopg://postgres@localhost:5432/postgres
%sql DROP DATABASE IF EXISTS exercise_3;
````
