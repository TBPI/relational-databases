# Exercise 37: Employees Whose Manager Left the Company

**Source:** [LeetCode 1978](https://leetcode.com/problems/employees-whose-manager-left-the-company/) (Easy).

Table: `Employees`
```
+-------------+----------+
| Column Name | Type     |
+-------------+----------+
| employee_id | int      |
| name        | varchar  |
| manager_id  | int      |
| salary      | int      |
+-------------+----------+
```

- `employee_id` is the primary key.
- `manager_id` is the `employee_id` of the employee's manager (`NULL` if they have none).
- When a manager leaves, their row is removed, but their former reports keep the old `manager_id`.

**Task.** List the ids of all employees who earn less than 30000 and whose manager has left the company, sorted by `employee_id`.

#### Fill the database with content

````python
%%sql
DROP TABLE IF EXISTS Employees;

CREATE TABLE Employees (
    employee_id INTEGER PRIMARY KEY,
    name VARCHAR NOT NULL,
    manager_id INTEGER,
    SALARY INTEGER NOT NULL
);

INSERT INTO Employees (employee_id, name, manager_id, salary)
VALUES
(3, 'Mila', 9, 60301),
(12, 'Antonella', NULL, 31000),
(13, 'Emery', NULL, 67084),
(1, 'Kalel', 11, 21241),
(9, 'Mikaela', NULL, 50937),
(11, 'Joziah', 6, 28485);

SELECT * FROM Employees;
````

#### Solve the exercise

**Task.**
List the ids of all employees who earn less than 30000 and whose manager has left the company, sorted by `employee_id`.

````python
%%sql
SELECT employee_id FROM Employees WHERE
(Employees.salary < 30000)
AND
(Employees.manager_id NOT IN (
    SELECT employee_id FROM Employees
))
ORDER BY employee_id;
````

#### Explanation

- The table should only return `employee_id`,
- The `WHERE` clause needs to conditions: first `salary` must be less than 30000, hence `Employees.salary <30000`.
- The second condition is connected via `AND`, and must ensure that the `manager_id` is not an `employee_id`.
The subquery `SELECT employee_id FROM Employees` returns all `employee_id`s.
- Note that `NULL` cannot be in a set in the sense of the `IN`-operation.
