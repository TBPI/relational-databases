# Exercise 9: Rising temperature

Source: [Leet Code 197](https://leetcode.com/problems/rising-temperature/description/?envType=study-plan-v2&envId=top-sql-50) (easy).

Table: `Weather`
```
+---------------+---------+
| Column Name   | Type    |
+---------------+---------+
| id            | int     |
| recordDate    | date    |
| temperature   | int     |
+---------------+---------+
```
- `id` is the column with unique values.
- `recordDate` is the date of the recorded temperature. There is at most one `temperature` measurement by day.

**Task.** Find all `id` as `Id` where the temperature is higher than the measured temperature the day before. Return the result in **any order**.

#### Fill the database with content

````python
%%sql
DROP TABLE IF EXISTS Weather;

CREATE TABLE Weather (
    id INTEGER UNIQUE NOT NULL,
    recordDate DATE UNIQUE NOT NULL,
    temperature INTEGER NOT NULL
);

INSERT INTO Weather (id, recordDate, temperature)
VALUES
(1, '2015-01-01', 10),
(2, '2015-01-02', 25),
(3, '2015-01-03', 20),
(4, '2015-01-04', 30);

SELECT * FROM Weather;
````

#### Solve the exercise

**Task.** Find all `id` as `Id` where the temperature is higher than the measured temperature the day before. Return the result in **any order**.

````python
%%sql
SELECT
    W.id AS Id
FROM
    Weather W INNER JOIN  Weather B
ON
    W.recordDate = B.recordDate::date + interval '1 day'
WHERE
    W.temperature > B.temperature;
````

#### Explanation

- Do an inner join on the such that every row contains the temperature of the present day and the day before (if existent).
- Then return `W.id` whenever the first temperature is higher than the one of yesterday (`W.temperature > B.temperature`).
- Note that PostgreSQL does not have the function `DATEADD(unit, value, date)`.
