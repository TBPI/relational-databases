#### Fill the database with content

````python
%%sql

DROP TABLE IF EXISTS Employee;

CREATE TABLE Employee (
    id INTEGER PRIMARY KEY,
    name VARCHAR NOT NULL,
    department VARCHAR,
    managerId INTEGER
);

INSERT INTO Employee (id, name, department, managerId)
VALUES
    (101, 'John', 'A', NULL),
    (102, 'Dan', 'A', 101),
    (103, 'James', 'A', 101),
    (104, 'Amy', 'A', 101),
    (105, 'Anne', 'A', 101),
    (106, 'Ron', 'B', 101);

SELECT * FROM Employee;
````

#### Solve the exercise

**Task.** Find the `name` of all managers who have **more than 4 reports**.
Return the  result in **any order**.

````python
%%sql
SELECT
    M.name FROM Employee M INNER JOIN (
        SELECT managerID AS id, COUNT(*) AS reports FROM Employee GROUP BY managerID
    ) R
ON
    M.id = R.id
WHERE R.reports > 4;
````

#### Explanation
- In a subquery create a table that counts the reports to each manager
- Perform an `INNER JOIN` with `Employee` on `Employee.id` with the `ID` of the manager.
- The `WHERE` clause ensure that there are at least 5 reports to the manager names which are returned.
