# Exercise 11: Employee Bonus

Table: `Employee`
```
+-------------+---------+
| Column Name | Type    |
+-------------+---------+
| empId       | int     |
| name        | varchar |
| supervisor  | int     |
| salary      | int     |
+-------------+---------+
```
- `empId` is the primary key.
- `name` is the name of the employee with `empId` and supervisor with id `supervisor`.
- `salary` denotes the employees salary.

Table: `Bonus`
```
+-------------+------+
| Column Name | Type |
+-------------+------+
| empId       | int  |
| bonus       | int  |
+-------------+------+
```
- `empID` is the foreign key from `Employee` and `bonus` the respective bonus.

**Task.**
Find all employees with `name` and `bonus` who did not receive any bonus or a bonus less than `1000`.

#### Fill the database with content

````python
%%sql

DROP TABLE IF EXISTS Employee;

CREATE TABLE Employee (
    empId INTEGER PRIMARY KEY,
    name VARCHAR NOT NULL,
    supervisor INTEGER,
    salary INTEGER
);

INSERT INTO Employee (empId, name, supervisor, salary)
VALUES
(3, 'Brad', NULL, 4000),
(1, 'John', 3, 1000),
(2, 'Dan', 3, 2000),
(4, 'Thomas', 3, 4000);

DROP TABLE IF EXISTS Bonus;

CREATE TABLE Bonus (
    empId INTEGER NOT NULL UNIQUE,
    bonus INTEGER,
    FOREIGN KEY (empId) REFERENCES Employee(empId)
);

INSERT INTO Bonus (empId, bonus)
VALUES
(2, 500),
(4, 2000);

SELECT * FROM Bonus;
````

#### Solve the exercise

**Task.**
Find all employees with `name` and `bonus` who did not receive any bonus or a bonus less than `1000`.

````python
%%sql
SELECT
    E.name, B.bonus
FROM
    Employee E LEFT JOIN Bonus B
ON
    E.empId = B.empId
WHERE
    B.bonus IS NULL OR B.bonus < 1000;
````

#### Explanation

- One needs to record every employee whether they appear in the `Bonus` table or not. Therefore do `Employee E LEFT JOIN Bonus B` on the `empId`.
- The two conditions on the bonus can be realised with a `WHERE` clause.
