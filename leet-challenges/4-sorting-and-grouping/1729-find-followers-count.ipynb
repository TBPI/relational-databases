# Exercise 27: Find Followers Count

**Source:** [LeetCode 1729](https://leetcode.com/problems/find-followers-count/description/?envType=study-plan-v2&envId=top-sql-50) (Easy).

Table: `Followers`
```
+-------------+------+
| Column Name | Type |
+-------------+------+
| user_id     | int  |
| follower_id | int  |
+-------------+------+
```
- `(user_id, follower_id)` is the primary key.

**Task.** For each `user_id` find the number of followers in ascending order for `user_id`.

Connect to the database `postgres` using the username `postgres`.

````python
%load_ext sql
%sql postgresql+psycopg://postgres@localhost:5432/postgres
````

### Create new database for the exercise

Make sure first that a database with the same name is removed in case it already exists.

````python
%sql DROP DATABASE IF EXISTS exercise_27 WITH (FORCE);
%sql CREATE DATABASE exercise_27;
````

Connect to the database.

*Caution.* No semicolon at the end otherwise it belongs to the name of the database.

````python
%sql postgresql+psycopg://postgres@localhost:5432/exercise_27 --alias exercise_27
````

#### Fill the database with content

````python
%%sql

CREATE TABLE IF NOT EXISTS Followers(
    user_id INTEGER,
    follower_id INTEGER
);

INSERT INTO Followers (user_id, follower_id) 
VALUES
(0, 1),
(1, 0),
(2, 0),
(2, 1);

SELECT * FROM Followers
````

#### Solve the exercise

**Task.** For each `user_id` find the number of followers in ascending order for `user_id`.

````python
%%sql
SELECT
    user_id, COUNT(*) AS followers_count
FROM
    Followers
GROUP BY 
    user_id
ORDER BY 
    user_id
````

#### Explanation
- Do a count over all rows with the same `user_id` in grouping by `user_id`.
- Don't forget to order by `user_id`.
