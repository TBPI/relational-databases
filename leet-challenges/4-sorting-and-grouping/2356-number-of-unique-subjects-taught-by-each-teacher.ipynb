# Exercise 23: Number of Unique Subjects Taught by Each Teacher

**Source:** [LeetCode 2356](https://leetcode.com/problems/number-of-unique-subjects-taught-by-each-teacher/) (Easy).

Table: `Teacher`
```
+-------------+------+
| Column Name | Type |
+-------------+------+
| teacher_id  | int  |
| subject_id  | int  |
| dept_id     | int  |
+-------------+------+
```

- `(subject_id, dept_id)` is the primary key.
- Each row says which teacher teaches a subject in a department, so a teacher can teach the same subject in several departments.

**Task.** For each teacher, count how many different subjects they teach, as a column `cnt`, in any order.

Connect to the database `postgres` using the username `postgres`.

````python
%load_ext sql
%sql postgresql+psycopg://postgres@localhost:5432/postgres
````

### Create new database for the exercise

````python
%sql DROP DATABASE IF EXISTS exercise_23 WITH (FORCE);
%sql CREATE DATABASE exercise_23;
````

Connect to the database.

````python
%sql postgresql+psycopg://postgres@localhost:5432/exercise_23 --alias exercise_23
````

#### Fill the database with content

````python
%%sql
DROP TABLE IF EXISTS Teacher;

CREATE TABLE Teacher (
    teacher_id INTEGER NOT NULL,
    subject_id INTEGER NOT NULL,
    dept_id INTEGER NOT NULL,
    PRIMARY KEY (subject_id, dept_id)
);

INSERT INTO Teacher (teacher_id, subject_id, dept_id)
VALUES
(1, 2, 3),
(1, 2, 4),
(1, 3, 3),
(2, 1, 1),
(2, 2, 1),
(2, 3, 1),
(2, 4, 1);

SELECT * FROM Teacher;
````

#### Solve the exercise

**Task.**
For each teacher, count how many different subjects they teach, as a column `cnt`, in any order.

````python
%sql SELECT teacher_id, COUNT(DISTINCT subject_id) AS cnt FROM Teacher GROUP BY teacher_id;
````

#### Explanation
- The Output must be the `teacher_id` and the number of *different* subjects this teacher teaches in a column called `cnt`.
- Use `GROUP BY teacher_id` to have one row for each id.
- Use `COUNT(DISTINCT ...)` to count all distinct `subject_id` within one `teacher_id`.
