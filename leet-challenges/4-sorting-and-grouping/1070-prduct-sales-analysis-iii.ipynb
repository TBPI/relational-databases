#### Fill the database with content

#### Solve the exercise

**Task.**
Find all sales for product during the first year it was sold (in **any order**).
Every queried row must contain `product_id`, `first_year`, `quantity` and `price`.

````python
%%sql
SELECT
    S.product_id, S.year AS first_year, S.quantity, S.price
FROM
    Sales S INNER JOIN (
        SELECT product_id, MIN(year) AS year FROM Sales GROUP BY product_id
    ) F
ON S.product_id = F.product_id AND S.year = F.year;
````

#### Explanation

- In a subquery select only the first year each product was sold using `MIN(YEAR)` grouping by `product_id`
- Then join this table on an `INNER JOIN` on `product_id` and `year`.
