# Exercise 47: Second Highest Salary

Table: `Employee`
```
+-------------+------+
| Column Name | Type |
+-------------+------+
| id          | int  |
| salary      | int  |
+-------------+------+
```
- `id` is the primary key.
- `salary` is the employee's salary with identity `id`.

**Task.**
Find the (strict) **second highest salary** as `SecondHighestSalary` and return `NULL` if it does not exist.

Connect to the database `postgres` using the username `postgres`.

````python
%load_ext sql
%sql postgresql+psycopg://postgres@localhost:5432/postgres
````

Connect to the database.

````python
%sql postgresql+psycopg://postgres@localhost:5432/exercise_47 --alias exercise_47
````

#### Fill the database with content

````python
%%sql
DROP TABLE IF EXISTS Employee;

CREATE TABLE Employee (
    id SERIAL PRIMARY KEY,
    salary INTEGER
);

INSERT INTO Employee (salary)
VALUES
(100),
(200),
(300);
````

#### Solve the exercise

**Task.**
Find the (strict) **second highest salary** as `SecondHighestSalary` and return `NULL` if it does not exist.

````python
%%sql
SELECT
    MAX(salary) AS SecondHighestSalary
FROM
    Employee
WHERE
    salary NOT IN (SELECT MAX(salary) FROM Employee);
````

#### Explanation
- Select the `MAX(salary)` from `Employee` which does not contain the `MAX(salary)` anymore.
