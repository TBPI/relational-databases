# Exercise 44: Fix Names in a Table

**Source:** [LeetCode 1667](https://leetcode.com/problems/fix-names-in-a-table/?envType=study-plan-v2&envId=top-sql-50)

Table: `Users`
```
+----------------+---------+
| Column Name    | Type    |
+----------------+---------+
| user_id        | int     |
| name           | varchar |
+----------------+---------+
```
- `user_id` (of the user) is the primary key.
- `name` is the name of the user.

**Task.** Change every name such that only the first letter is uppercase and all others are lowercase, e.g. `mAriE ANN` $\to$ `Marie ann`.
Return the table **ordered by** `user_id`.

Connect to the database `postgres` using the username `postgres`.

````python
%load_ext sql
%sql postgresql+psycopg://postgres@localhost:5432/postgres
````

### Create new database for the exercise

Make sure first that a database with the same name is removed in case it already exists.

````python
%sql DROP DATABASE IF EXISTS exercise_44 WITH (FORCE);
%sql CREATE DATABASE exercise_44;
````

Connect to the database.

*Caution.* No semicolon at the end otherwise it belongs to the name of the database.

````python
%sql postgresql+psycopg://postgres@localhost:5432/exercise_44 --alias exercise_44
````

#### Fill the database with content

````python
%%sql
DROP TABLE IF EXISTS Users;

CREATE TABLE Users (
    user_id INTEGER PRIMARY KEY,
    name VARCHAR
);

INSERT INTO Users (user_id, name)
VALUES
(1, 'aLice'),
(2, 'bOB'),
(3, 'mAriE ANN');

SELECT * FROM Users;
````

#### Solve the exercise

**Task.** Change every name such that only the first letter is uppercase and all others are lowercase, e.g. `mAriE ANN` $\to$ `Marie ann`.
Return the table **ordered by** `user_id`.

````python
%sql SELECT user_id, UPPER(LEFT(name, 1)) || LOWER(RIGHT(name, LENGTH(name)-1)) AS name FROM Users ORDER BY user_id;
````

#### Explanation
- Output must be in ascending order by `user_id`, hence the query finishes with `ORDER BY user_id`.
- The first letter of `name` must be uppercase. Taking the first $n$ letters can be done with `LEFT( _, n)`. Then `UPPER` sets it to uppercase.
- All the remaining letters must be lower case. This can be achieved using `RIGHT( _ , LENGTH( _ ) - 1)` and then `LOWER`.
- Two concatenate the two strings use `||`.

#### Close connection
