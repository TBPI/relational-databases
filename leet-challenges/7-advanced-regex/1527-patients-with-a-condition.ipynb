# Exercise 45: Patients With a Condition

**Source:** [LeetCode 1527](https://leetcode.com/problems/patients-with-a-condition/?envType=study-plan-v2&envId=top-sql-50) (Easy).

Table: `Patients`
```
+--------------+---------+
| Column Name  | Type    |
+--------------+---------+
| patient_id   | int     |
| patient_name | varchar |
| conditions   | varchar |
+--------------+---------+
```
- `patient_id` is the primary key.
- `conditions` contains information about the patient `patient_name`.

**Task.**
Find the `patient_id`, `patient_name` and the conditions of patients who have Type I Diabetes (starts with `DIAB1`).

Connect to the database `postgres` using the username `postgres`.

````python
%load_ext sql
%sql postgresql+psycopg://postgres@localhost:5432/postgres
````

### Create new database for the exercise

Make sure first that a database with the same name is removed in case it already exists.

````python
%sql DROP DATABASE IF EXISTS exercise_45 WITH (FORCE);
%sql CREATE DATABASE exercise_45;
````

Connect to the database.

*Caution.* No semicolon at the end otherwise it belongs to the name of the database.

````python
%sql postgresql+psycopg://postgres@localhost:5432/exercise_45 --alias exercise_45
````

#### Fill the database with content

````python
%%sql
DROP TABLE IF EXISTS Patients;

CREATE TABLE Patients (
    patient_id INTEGER PRIMARY KEY,
    patient_name VARCHAR NOT NULL,
    conditions VARCHAR
);

INSERT INTO Patients (patient_id, patient_name, conditions)
VALUES
(1, 'Daniel', 'YFEV COUGH'),
(2, 'Alice', NULL),
(3, 'Bob', 'DIAB100 MYOP'),
(4, 'George', 'ACNE DIAB100'),
(5, 'Alain', 'DIAB201');

SELECT * FROM Patients;
````

#### Solve the exercise

**Task.**
Find the `patient_id`, `patient_name` and the conditions of patients who have Type I Diabetes (starts with `DIAB1`).

````python
%%sql
SELECT
    patient_id, patient_name, conditions
FROM
    Patients
WHERE
    conditions LIKE 'DIAB1%' OR conditions LIKE '% DIAB1%';
````

#### Explanation

- The conditions column needs to start with `DIAB1` or ` DIAB1` (as there might have been another condition before). Afterwards, anything can be written, therefore the `%`.
