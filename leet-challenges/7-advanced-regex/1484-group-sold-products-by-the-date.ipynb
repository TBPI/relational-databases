# Exercise 48: Group Sold Products By The Date

**Source:** [LeetCode 1484](https://leetcode.com/problems/group-sold-products-by-the-date/description/?envType=study-plan-v2&envId=top-sql-50) (Easy).

Table: `Activities`
```
+-------------+---------+
| Column Name | Type    |
+-------------+---------+
| sell_date   | date    |
| product     | varchar |
+-------------+---------+
```
- No primary key, i.e. the table may contain duplicates.
- `sell_date` is the date a `product` is sold.

**Task.**
Find for each date the number of different products and their names (no duplicates) ordered by `sell_date`.

#### Fill the database with content

````python
%%sql
DROP TABLE IF EXISTS Activities;

CREATE TABLE Activities (
    sell_date DATE,
    product VARCHAR
);

INSERT INTO Activities (sell_date, product)
VALUES
('2020-05-30', 'Headphone'),
('2020-06-01', 'Pencil'),
('2020-06-02', 'Mask'),
('2020-05-30', 'Basketball'),
('2020-06-01', 'Bible'),
('2020-06-02', 'Mask'),
('2020-05-30', 'T-Shirt');
````

#### Solve the exercise

**Task.**
Find for each date the number of different products and their names (no duplicates) ordered by `sell_date`.

````python
%%sql
SELECT
    sell_date, COUNT(DISTINCT product) AS num_sold, STRING_AGG(DISTINCT product, ',' ORDER BY product) AS products
FROM
    Activities
GROUP BY
    sell_date
ORDER BY
    sell_date
````

#### Explanation

- Group by `sell_date` (and order by it in the end).
- Count the number of distinct products sold using `COUNT(DISTINCT product)`.
- `STRING_AGG(DISTINCT _ , ',' ORDER BY product)` concatenates all the products belonging to the group using a comma. The `DISTINCT`, and `ORDER BY product` ensure no duplicates and that the list is in lexicographic order.
