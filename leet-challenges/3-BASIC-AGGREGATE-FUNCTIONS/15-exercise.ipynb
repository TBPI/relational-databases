```
+----------------+----------+
| Column Name    | Type     |
+----------------+----------+
| id             | int      |
| movie          | varchar  |
| description    | varchar  |

+----------------+----------+
```

- `id` is the primary key (column with unique values),
- `name` is the name of the movie,
- `description` is the genre of the movie,
- `rating` is its rating between $[0,10]$ and with one decimal.

**Exercise.**
Write a solution to report the movies with an **odd-numbered ID** and a description that is not `boring`.

Return the result table **ordered by rating in descending order**.

#### Fill the database with content

````python
%%sql
DROP TABLE IF EXISTS Cinema;

CREATE TABLE Cinema (
    id INTEGER PRIMARY KEY,
    movie VARCHAR NOT NULL,
    description VARCHAR,
    rating FLOAT
);

INSERT INTO Cinema (id, movie, description, rating)
VALUES
(1, 'War', 'great 3D', 8.9),
(2, 'Science', 'fiction', 8.5),
(3, 'irish', 'boring', 6.2),
(4, 'Ice song', 'Fantacy', 8.6),
(5, 'House card', 'Interesting', 9.1);

SELECT * FROM Cinema;
````

#### Solve the exercise

**Exercise.**
Write a solution to report the movies with an **odd-numbered ID** and a description that is not `boring`.

Return the result table **ordered by rating in descending order**.

````python
%%sql
SELECT * FROM Cinema
WHERE
description != 'boring'
AND
MOD(id,2) = 1 ORDER BY rating DESC;
````

#### Explanation

- Each feature must be selected for the corresponding row, hence `SELECT * FROM ...`,
- There are two conditions which need both to hold, hence `AND`,
- `description` must not be boring,
- `id` must be odd, i.e. $ = 1\mod 2$ (use `MOD` function).
